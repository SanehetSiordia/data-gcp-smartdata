```python
import pandas as pd
from google.cloud import storage
```

```python
client = storage.Client(project='smart-data-clase-06')
bucket_name = 'sd-gcolab-bucket-2'
bucket = client.create_bucket(bucket_name)
```

```python
!pwd
```

```python
columns = ['order_id', 'order_date', 'order_customer_id', 'order_status']
```

Subimos el archivo orders.csv o cualquier archivo csv que tengamos a la mano.

```python
df = pd.read_csv('orders.csv', names=columns)
df
```

```python
df.to_csv(
    'gs://sd-gcolab-bucket-2/orders.csv',
    index=False
)
```

```python
df.to_parquet(
    'gs://sd-gcolab-bucket-2/orders.parquet',
    index=False
)
```

```python
df_parquet = pd.read_parquet('gs://sd-gcolab-bucket-2/orders.parquet')
df_parquet
```
