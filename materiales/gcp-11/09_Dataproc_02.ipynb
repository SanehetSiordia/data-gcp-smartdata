**create_cluster():** Este método se utiliza para crear un clúster de Dataproc. Ejemplo de cómo crear un clúster de 2 nodos:

```python
from google.cloud import dataproc_v1
from google.cloud.dataproc_v1.types import ClusterConfig, InstanceGroupConfig, GceClusterConfig

# Endpoint para la región us-central1
endpoint = "us-central1-dataproc.googleapis.com:443"

# Crea un cliente con el endpoint específico de la región
client = dataproc_v1.ClusterControllerClient(client_options={"api_endpoint": endpoint})

cluster_config = ClusterConfig(
    gce_cluster_config = GceClusterConfig(zone_uri='us-central1-a'),

    master_config=InstanceGroupConfig(
        num_instances=1,
        machine_type_uri='n1-standard-2',
        disk_config={'boot_disk_type': 'pd-standard', 'boot_disk_size_gb': 100}
    ),
    worker_config=InstanceGroupConfig(
        num_instances=2,
        machine_type_uri='n1-standard-2',
        disk_config={'boot_disk_type': 'pd-standard', 'boot_disk_size_gb': 100}
    )
)


# Initialize request argument(s)
cluster = dataproc_v1.Cluster(
    project_id="neural-truth-406011",
    cluster_name="ronny-cluster",
    config=cluster_config
)

request = dataproc_v1.CreateClusterRequest(
    project_id="neural-truth-406011",
    region="us-central1",
    cluster=cluster
)

# Make the request
operation = client.create_cluster(request=request)

print("Waiting for operation to complete...")

response = operation.result()

# Handle the response
print(response)
```

**list_clusters():** Este método se utiliza para listar los clústeres de Dataproc en un proyecto. Aquí hay un ejemplo de cómo listar los clústeres:

```python
from google.cloud import dataproc_v1

project_id = 'neural-truth-406011'
region = 'us-central1'

client = dataproc_v1.ClusterControllerClient(client_options={"api_endpoint": endpoint})

result = client.list_clusters(
    project_id=project_id,
    region=region
)

for cluster in result.clusters:
    print(cluster.cluster_name)
```

**get_cluster():** Este método se utiliza para obtener información sobre un clúster de Dataproc. Aquí hay un ejemplo de cómo obtener información sobre un clúster:

```python
from google.cloud import dataproc_v1

project_id = 'neural-truth-406011'
region = 'us-central1'
cluster_name = 'ronny-cluster'

client = dataproc_v1.ClusterControllerClient(client_options={"api_endpoint": endpoint})

result = client.get_cluster(
    project_id=project_id,
    region=region,
    cluster_name=cluster_name
)

print(result)
```

**submit_job():** Este método se utiliza para enviar un trabajo a un clúster de Dataproc. Aquí hay un ejemplo de cómo enviar un trabajo de PySpark:

```python
from google.cloud import dataproc_v1
from google.cloud.dataproc_v1.types import Job, PySparkJob, SparkJob, JobPlacement

project_id = 'neural-truth-406011'
region = 'us-central1'
cluster_name = 'ronny-cluster'

client = dataproc_v1.JobControllerClient(client_options={"api_endpoint": endpoint})

job = Job(
    placement=JobPlacement(
        cluster_name=cluster_name
    ),
    pyspark_job=PySparkJob(
        main_python_file_uri='gs://dev-ronny-datalake-gcp/resources/transformaciones_pyspark.py',
        jar_file_uris = ['gs://dev-ronny-datalake-gcp/resources/spark-bigquery-with-dependencies_2.12-0.30.0.jar']
    )
)

operation = client.submit_job(
    project_id=project_id,
    region=region,
    job=job
)
result = operation.job_uuid
print(result)
```

**delete_cluster():** Este método se utiliza para eliminar un clúster de Dataproc. Aquí hay un ejemplo de cómo eliminar un clúster:

```python
from google.cloud import dataproc_v1

project_id = 'neural-truth-406011'
region = 'us-central1'
cluster_name = 'ronny-cluster'

client = dataproc_v1.ClusterControllerClient(client_options={"api_endpoint": endpoint})

operation = client.delete_cluster(
    project_id=project_id,
    region=region,
    cluster_name=cluster_name
)

result = operation.result()
print(result)
```
